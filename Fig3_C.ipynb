```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
fig, ax = plt.subplots()

fruits = pd.concat([biologicalProcess.BiologicalProcess[0:5],molecularFunction.MolecularFunction[0:5],cellularComponent.CellularComponent[0:5]])[::-1]
counts = pd.concat([biologicalProcess.ProteinCount[0:5], molecularFunction.ProteinCount[0:5], cellularComponent.ProteinCount[0:5]])[::-1]
bar_labels = pd.concat([biologicalProcess.BiologicalProcess[0:5], molecularFunction.MolecularFunction[0:5], cellularComponent.CellularComponent[0:5]])[::-1]
bar_colors = ['tab:red','tab:blue','tab:green'][::-1]

import itertools

k = 5

ax.barh(fruits, counts, label=bar_labels, color=list(itertools.chain(*[itertools.repeat(i, k) for i in bar_colors])))

#ax.set_ylabel('Top 10 GO:Term connected to proteins ')
ax.set_title('GO:Terms connected to potential marker metaproteins')
ax.get_xaxis().set_major_formatter(
    matplotlib.ticker.FuncFormatter(lambda counts, p: format(int(counts), ',')))
red_patch = mpatches.Patch(color='tab:red', label='Biological Process')
blue_patch = mpatches.Patch(color='tab:blue', label='Molecular Function')
green_patch = mpatches.Patch(color='tab:green', label='Cellular Component')

plt.xticks(fontsize= 13)
plt.yticks(fontsize= 13)
plt.legend(handles=[red_patch, blue_patch, green_patch], fontsize = 11)
#ax.legend(title='Fruit color')
plt.savefig("C:/Users/max-w/Desktop/Top5_GO_Terms", dpi=300, bbox_inches='tight')
plt.show()
```
